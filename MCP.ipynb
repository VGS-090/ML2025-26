```python
def mcp_neuron(inputs, weights, threshold):
    # Calculate weighted sum
    weighted_sum = sum(i * w for i, w in zip(inputs, weights))
    # Activation function (step function)
    return 1 if weighted_sum >= threshold else 0

# AND gate
def AND_gate(x1, x2):
    weights = [1, 1]
    threshold = 2
    return mcp_neuron([x1, x2], weights, threshold)

# NOT gate
def NOT_gate(x):
    weights = [-1]
    threshold = 0
    return mcp_neuron([x], weights, threshold)

# Test all input combinations for AND gate
print("AND Gate")
for x1 in [0, 1]:
    for x2 in [0, 1]:
        print(f"AND({x1}, {x2}) = {AND_gate(x1, x2)}")

# Test input for NOT gate
print("\nNOT Gate")
for x in [0, 1]:
    print(f"NOT({x}) = {NOT_gate(x)}")
```

```python
def mcp_neuron(inputs, weights, threshold):
    weighted_sum = sum(i * w for i, w in zip(inputs, weights))
    return 1 if weighted_sum >= threshold else 0

def AND_gate(x1, x2):
    weights = [1, 1]
    threshold = 2
    return mcp_neuron([x1, x2], weights, threshold)

def OR_gate(x1, x2):
    weights = [1, 1]
    threshold = 1
    return mcp_neuron([x1, x2], weights, threshold)

def get_binary_input(prompt):
    while True:
        try:
            value = int(input(prompt))
            if value in (0, 1):
                return value
            else:
                print("Please enter 0 or 1.")
        except ValueError:
            print("Invalid input. Please enter 0 or 1.")

print("MCP Neuron Simulation for AND and OR gates")

gate_choice = input("Choose gate (AND/OR): ").strip().upper()

if gate_choice == "AND":
    x1 = get_binary_input("Enter input x1 (0 or 1): ")
    x2 = get_binary_input("Enter input x2 (0 or 1): ")
    output = AND_gate(x1, x2)
    print(f"AND({x1}, {x2}) = {output}")

elif gate_choice == "OR":
    x1 = get_binary_input("Enter input x1 (0 or 1): ")
    x2 = get_binary_input("Enter input x2 (0 or 1): ")
    output = OR_gate(x1, x2)
    print(f"OR({x1}, {x2}) = {output}")

else:
    print("Invalid gate choice. Please enter 'AND' or 'OR'.")
```

```python
def mcp_neuron(inputs, weights, threshold):
    weighted_sum = sum(i * w for i, w in zip(inputs, weights))
    return 1 if weighted_sum >= threshold else 0

def AND_gate(x1, x2):
    return mcp_neuron([x1, x2], [1, 1], 2)

def OR_gate(x1, x2):
    return mcp_neuron([x1, x2], [1, 1], 1)

def NOT_gate(x):
    return mcp_neuron([x], [-1], 0)

def XOR_gate(x1, x2):
    or_result = OR_gate(x1, x2)
    and_result = AND_gate(x1, x2)
    not_and = NOT_gate(and_result)
    return AND_gate(or_result, not_and)

# Test XOR gate
for x1 in [0, 1]:
    for x2 in [0, 1]:
        print(f"XOR({x1}, {x2}) = {XOR_gate(x1, x2)}")
```
